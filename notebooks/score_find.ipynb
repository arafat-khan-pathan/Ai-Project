```python
import pandas as pd
import pickle

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import sys
import os

sys.path.append(os.path.abspath(".."))

from src.nlp_preprocessing import preprocess_text
```

```python
# 1. Load test data
test_df = pd.read_csv(
    "../dataset/test-ham-spam.csv",
    encoding="latin-1"
)

test_df.columns = ["label", "message"]

# Remove empty values
test_df = test_df.dropna()
test_df = test_df[test_df["message"].str.strip() != ""]
```

```python
# 2. Convert labels
# ham = 0
# spam = 1
test_df["label"] = test_df["label"].map({
    "ham": 0,
    "spam": 1
})
```

```python
# 3. Preprocess test messages
test_df["cleaned_message"] = test_df["message"].apply(
    preprocess_text
)
```

```python
# 4. Load trained TF-IDF vectorizer
with open("../models/tfidf_vectorizer.pkl", "rb") as f:
    vectorizer = pickle.load(f)
```

```python
# 5. Load trained Random Forest model
with open("../models/random_forest.pkl", "rb") as f:
    model = pickle.load(f)
```

```python
# 6. Convert test messages to TF-IDF
X_test_tfidf = vectorizer.transform(
    test_df["cleaned_message"]
)
y_test = test_df["label"]
```

```python
# 7. Predict
y_pred = model.predict(X_test_tfidf)
```

```python
# 8. Calculate metrics
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    pos_label=1
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label=1
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label=1
)
```

```python
# 9. Print results
print("========== MODEL PERFORMANCE ==========")

print(f"Accuracy : {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall   : {recall * 100:.2f}%")
print(f"F1-Score : {f1 * 100:.2f}%")
```
