```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# 1. Load original dataset
df = pd.read_csv("spam.csv", encoding="latin-1")

# Keep only required columns
df = df[["v1", "v2"]]

# Rename columns
df.columns = ["label", "message"]

print("Column Names:")
print(df.columns.tolist())

print("\nTotal Rows:")
print(len(df))

print("\nShape:")
print(df.shape)
```

```python
# 2. Separate Ham and Spam
ham_df = df[df["label"] == "ham"].copy()
spam_df = df[df["label"] == "spam"].copy()
```

```python
# 3. Save all Ham and Spam
ham_df.to_csv("sep_ham.csv", index=False)
spam_df.to_csv("sep_spam.csv", index=False)
```

```python
# 4. Ham → 80% Train, 20% Test
ham_train, ham_test = train_test_split(
    ham_df,
    test_size=0.20,
    random_state=42
)


# 5. Spam → 80% Train, 20% Test
spam_train, spam_test = train_test_split(
    spam_df,
    test_size=0.20,
    random_state=42
)
```

```python
# 6. Create Train Dataset
train_df = pd.concat([
    ham_train,
    spam_train
])

# 7. Create Test Dataset
test_df = pd.concat([
    ham_test,
    spam_test
])


# Shuffle train data
train_df = train_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)


# Shuffle test data
test_df = test_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
# 8. Save Train & Test
train_df.to_csv("train.csv", index=False)

test_df.to_csv("test-ham-spam.csv",index=False)
```

```python
# 9. Show information
print("========== Dataset ==========")
print("Total:", len(df))
print("Ham:", len(ham_df))
print("Spam:", len(spam_df))

print("\n========== Train ==========")
print("Train total:", len(train_df))
print(train_df["label"].value_counts())

print("\n========== Test ==========")
print("Test total:", len(test_df))
print(test_df["label"].value_counts())

print("\n========== Files Created ==========")
print("1. sep_ham.csv")
print("2. sep_spam.csv")
print("3. train.csv")
print("4. test-ham-spam.csv")
```
